#### 1. compute the frequency of the letters used in the texts (you may want to use `ascii_lowercase` from the module `string`)

````python
import string 
def letters_occurences(lecture:str) -> dict :
    """
    The argument lecture have to be a string
    The code will return a dictionnary with the occurences of the letters in the string
    
    Examples: 
        letters_occurences('') returns {}
        letters_occurences('ab') returns {'a': 1, 'b': 1}
    """
    #we create a dictionnary for the frequencies of the letters used in the text
    occurences_letter={} 

    for item in lecture: 
        # we check if the character is in the ascii_lowercase to sort letters from non-letters
        if item in string.ascii_lowercase:
            # if the letter don't have an associate key in the dictionnary
            if item not in occurences_letter:
                #we create a key corresponding to the letter and we start the counter with one
                occurences_letter[item]=1
            # if the letter have an associate key in the dictionnary
            else:
                #we add one to the counter associated to the key
                occurences_letter[item]+=1
    return occurences_letter
letters_occurences(lecture)
````

#### 2. compute the frequency of the words in the text(s) (you may want to use `punctuation` from the module `string`, `parse` from the module `russel.py`);

````python
from russell import parse
def take_out_non_letter(lecture:str)-> str:
    """
    The argument lecture have to be a string
    The code will return a string containing only letters and space
    
    Examples: 
        letters_frequency('') returns ''
        letters_frequency('a.b') returns 'ab'
        letters_frequency('a, ui') returns 'ab ui'
    """ 
    #I give to parse a string and the characters I'm looking for inside the given string it gives me back a list of items 
    # a item is either a string containing only letter and space, or a string containing one character of non-letter
    lecture_splitting_punctuation=parse(lecture,string.punctuation) 
    #we gonna build a new string whitout the non-letter
    lecture_whithout_punctuation=''
    #for each element of lecture_splitting_punctuation
    for item in lecture_splitting_punctuation :
        #I check if the element is not non-letter character
        if item not in string.punctuation:
            #I add this element to the new-built string
            lecture_whithout_punctuation+=item
    return lecture_whithout_punctuation

def frequency_words(lecture : str) -> dict :
    """
    The argument lecture have to be a string
    The code will return a dict with the frequency of the words in the string
    
    Examples: 
        letters_frequency('') returns {}
        letters_frequency('ab') returns {'ab' : 1}
        letters_frequency('a, ui') returns {'ab' : 1, 'ui': 1}
    """
    #will return a string containing only letters and space
    cleaned_lecture=take_out_non_letter(lecture)
    #we create a dictionnary for the frequencies of the word used in the text 
    frequencies_word={}
    #splits lecture_whitout_punctuation into a list with the whitespace as separator, each item is a word
    for item in cleaned_lecture.split():
        #I check if word is already registered in frequencies_word dictionnary
        if item not in frequencies_word:
            #if not, I register it and set its counter to one
            frequencies_word[item]=1
        # if yes
        else:
            #I add one to its counter
            frequencies_word[item]+=1
    return frequencies_word   
frequency_words(lecture)
````

#### 3. represent word transitions by dictionaries within a dictionary named transitions that will be used as `transitions[word[n]][word[n+1]] = count` where `count` represents the number of occurrences

````python
def word_sequences(lecture:str)-> dict:
    """
    The argument lecture have to be a string
    The code will return dictionnaries whithin a dictionary which represent the word transitions in the string lecture
    
    Examples: 
        word_sequences('') returns {None:{}}
        word_sequences('ab') returns {None:{'ab': 1}, 'ab':{None: 1}}
        word_sequences('a, ui') returns {None:{'a': 1}, 'a':{'ui': 1}, 'ui':{None: 1}}
    """
    #create the dictionnary
    transitions={}
    #create the key None
    transitions[None]={}

    # take out the punctuation and we use split() to make the list iterable
    item=take_out_non_letter(lecture).split()
    #to go through the list
    for k in range(0,len(item)) :
        #if a word do not have an associate key in the dictionnary
        if item[k] not in transitions:
            #we create one and we give it a dictionnary 
            transitions[item[k]]={}
    #to go through the list
    for i in range(0,len(item)):
        #if we are at the beginning at the list, the current word if mandatory following a None 
        if i==0:
            #if the word not have an associate key in the dictionnary transitions[None]
            if item[i] not in transitions[None]:
                #we create one and set its counter to one
                transitions[None][item[i]]=1
            #if yes he does 
            else:
                #we add one to its counter
                transitions[None][item[i]]+=1
        #if we are neither at the beginning nor the end
        if 0<=i<len(item)-1:
            #if the word following liste[i] not have an associate key in the dictionnary transitions[liste[i]]
            if item[i+1] not in transitions[item[i]]:
                #we create one and set its counter to one
                transitions[item[i]][item[i+1]]=1
            #if yes he does
            else:
                #we add one to its counter
                transitions[item[i]][item[i+1]]+=1
        #if we are at the end at the list, the following item if mandatory following a None 
        else:
            #if the word not have an associate key in the dictionnary transitions[liste[len(liste)]]
            if None not in transitions[item[i]]:
                #we create one and set its counter to one
                transitions[item[i]][None]=1
            #if yes he does
            else:
                #we add one to its counter
                transitions[item[i]][None]+=1
    return transitions
word_sequences(lecture)
````

#### 4. compute the probability of word transitions Pr(word[n+1] | word[n])

````python
def probability (lecture:str):
    """
    The argument lecture have to be a string
    The code will return the probability of word transitions Pr(word[n+1] | word[n]) based on the given lecture
    
    Examples: 
        word_sequences('') returns None
        word_sequences('ab') returns Pr( a | None )= 1.0 \n Pr( None | a )= 1.0
        word_sequences('a, ui') returns Pr( a | None )= 1.0 \n Pr( ui | a )= 1.0 \n Pr( None | ui )= 1.0
    """
    #assigne the dictionnaries whithin a dictionary which represent the word transitions in the string lecture to a variable
    d=word_sequences(lecture)
    #for each keys items (so the word[n]) included in transitions
    for wordn in d.keys():
        #for each keys items (so the word[n+1]) included in transitions[wordn]
        for wordn1 in d[wordn].keys():
            #the probability is the number of time wordn1 followed wordn divided by the number of time wordn was followed by a word
            prvalue=d[wordn][wordn1]/len(d[wordn1].keys())
            #we print the result
            print("Pr(",wordn1, "|",wordn, ")=", prvalue)
probability(lecture)
````

````python
from russell import translate_roman
def get_all_russel() -> str :
    """
    This function don't take any argument
    The code will return a string corresponding to all the russel texts
    
    Examples: 
        get_all_russel() returns 'there are...exists.'
    """
    all_text=[]
    all_lectures=''
    for i in range(1,16):
        all_text+=get_lectures()[0][translate_roman(i)]['text']
    for k in range(0,len(all_text)):
        all_lectures+=' '+ all_text[k] #we add a space before the item if not ['...dog']+['The ...] will give ['...dogThe...]
    return all_lectures.lower()
all_lecture=get_all_russel()
all_lecture
````

````python
from russell import parse
def text_cleaner(lecture:str) -> str:
    """
    The argument have to be a string
    The code will return a string containing only word, dealing with all special cases
    
    Examples:
        text_cleaner('') returns ''
        text_cleaner('According to the chapter alpha (p.24 ii)') returns 'According to the chapter alpha'
    """
    #I give to parse a string and the characters I'm looking for inside the given string and it t gives me back a list of items
    #an item is either a string containing only letter and space, or a string containing one character researched
    uncleaned_lecture=parse(lecture, string.punctuation +'0123456789'+'—')
    #create an empty string to build a string containing only word
    cleaned_lecture=''
    #create a parameter to know if the item we are looking at is between parenthesis or not
    into_parenthesis=False

    #we are going trough every item in uncleaned_lecture
    for item in uncleaned_lecture: 
        #parenthesis is opened
        if item =='(':
           into_parenthesis=True
        #parenthesis is closed
        if item==')':
            into_parenthesis=False
        #check if the item is not a researched character and if it's not between parenthensis
        if item not in string.punctuation+'0123456789'+'—' and into_parenthesis==False:
            cleaned_lecture +=item 
    return cleaned_lecture
text_cleaner(all_lecture)
````

### III. make your own Russell sentences based on the computed `transitions` where the sentence would be constructed by _sampling_ a next word based on the relative counts given the current word (see also the caseine exercise as an example)
